$$
 \boxed{V_{eff} = \dfrac{\ell^2}{2r^2} - \dfrac{r_s}{2r} - \dfrac{r_s\ell^2}{2r^3}.} 
$$
Use the radial geodesic-like equation which is given by: 
$$
\boxed{\dfrac{1}{2} \left( \dfrac{dr}{ds} \right)^2 + V_{eff} = E,} 
$$  
compute the trajectory between $r=r_s$ to $r\to \infty$ considering the initial condition
$$
\boxed{r(0) = 10\;r_s \;\; \& \;\; \dfrac{dr}{ds}(0) = 0.1}
$$
for different values of $\ell$ and $r_s$. Find possible values for circular motion.

**`Note : the allowed region for a trajectory must satisfy:`** $E - V_{eff} \geq 0$
